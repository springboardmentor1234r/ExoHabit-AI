```python
# Cell 1: imports and global config
import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix
from sklearn.utils import class_weight
import joblib
import shap

# Paths and constants
DATA_PATH = r"D:\Exo_Habit-AI\ExoPlanet.csv"
ARTIFACTS_DIR = r"D:\Exo_Habit-AI\artifacts"
os.makedirs(ARTIFACTS_DIR, exist_ok=True)
RANDOM_STATE = 42
```

```python
# Cell 2: load dataset and normalize column names
df = pd.read_csv(DATA_PATH)
# Normalize column names: strip, lowercase, replace spaces and special chars with underscore
def normalize_col(c):
    return (c.strip().lower()
              .replace(' ', '_')
              .replace('-', '_')
              .replace('.', '_')
              .replace('(', '')
              .replace(')', ''))
df.columns = [normalize_col(c) for c in df.columns]
print("Loaded dataset shape:", df.shape)
print("Columns:", df.columns.tolist()[:50])
df.head(7)
```

```python
# Cell 3: schema validation and quick inspection
def schema_report(df):
    report = pd.DataFrame({
        'dtype': df.dtypes.astype(str),
        'n_missing': df.isna().sum(),
        'pct_missing': (df.isna().mean()).round(4),
        'n_unique': df.nunique(dropna=False)
    })
    return report

schema = schema_report(df)
display(schema)
print("Sample rows:")
display(df.head(5))
print("If column names differ from expected, update mapping in Cell 4 mapping dictionary.")
```

```python
# Cell 4: canonical mapping of alternate column names to expected canonical names
canonical_map = {
    # identifiers
    'pl_name': 'planet_name', 'name': 'planet_name', 'planet': 'planet_name',
    # radius
    'pl_rade': 'radius', 'radius_earth': 'radius', 'radius': 'radius', 'r': 'radius',
    # mass
    'pl_bmasse': 'mass', 'mass_earth': 'mass', 'mass': 'mass', 'm': 'mass',
    # density
    'pl_dens': 'density', 'density': 'density',
    # equilibrium temperature
    'pl_eqt': 'equilibrium_temperature', 'teq': 'equilibrium_temperature', 'equilibrium_temperature': 'equilibrium_temperature',
    # orbital period
    'pl_orbper': 'orbital_period', 'orbital_period': 'orbital_period', 'period': 'orbital_period',
    # semi-major axis
    'pl_orbsmax': 'semi_major_axis', 'semi_major_axis': 'semi_major_axis', 'a': 'semi_major_axis',
    # distance
    'sy_dist': 'distance', 'distance': 'distance', 'pl_dis': 'distance',
    # host star properties
    'st_teff': 'host_star_temperature', 'host_star_temperature': 'host_star_temperature', 'teff': 'host_star_temperature',
    'st_lum': 'host_star_luminosity', 'host_star_luminosity': 'host_star_luminosity', 'luminosity': 'host_star_luminosity',
    'st_met': 'metallicity', 'metallicity': 'metallicity', 'fe_h': 'metallicity',
    'st_spectype': 'host_star_type', 'host_star_type': 'host_star_type', 'spectral_type': 'host_star_type'
}

# Build mapping for existing columns
rename_dict = {}
for col in df.columns:
    low = col.lower()
    if low in canonical_map:
        target = canonical_map[low]
        # avoid overwriting an existing canonical column
        if target in df.columns and target != col:
            # skip rename to avoid collision
            continue
        rename_dict[col] = target

if rename_dict:
    df = df.rename(columns=rename_dict)
    print("Applied renames:", rename_dict)
else:
    print("No automatic renames applied. If your dataset uses different names, add them to canonical_map.")
print("Columns after rename:", df.columns.tolist()[:50])
```

```python
# Cell 5: detect numeric-like columns and coerce them to numeric
def detect_numeric_like(df, threshold=0.6):
    numeric_candidates = []
    for c in df.columns:
        if c in ['planet_name', 'name', 'id', 'pl_name']:
            continue
        s = df[c].astype(str).fillna('')
        # remove common non-numeric characters but keep scientific notation
        cleaned = s.str.replace(r'[^0-9eE\+\-\.]', '', regex=True)
        coerced = pd.to_numeric(cleaned, errors='coerce')
        frac_numeric = coerced.notna().mean()
        if frac_numeric >= threshold:
            numeric_candidates.append(c)
    return numeric_candidates

numeric_like = detect_numeric_like(df, threshold=0.5)
print("Numeric-like columns detected:", numeric_like)

# Coerce detected numeric-like columns
for c in numeric_like:
    df[c] = pd.to_numeric(df[c].astype(str).str.replace(r'[^0-9eE\+\-\.]', '', regex=True), errors='coerce')

# Build expected_numeric list from canonical names present
expected_numeric_all = [
    'radius', 'mass', 'density', 'equilibrium_temperature',
    'orbital_period', 'semi_major_axis', 'distance',
    'host_star_temperature', 'host_star_luminosity', 'metallicity'
]
expected_numeric = [c for c in expected_numeric_all if c in df.columns]
missing_expected = [c for c in expected_numeric_all if c not in df.columns]
print("Canonical expected numeric present:", expected_numeric)
if missing_expected:
    print("Missing canonical numeric columns:", missing_expected)
# Fallback numeric candidates for feature engineering
fallback_numeric = [c for c in numeric_like if c not in expected_numeric]
print("Fallback numeric candidates:", fallback_numeric)
```

```python
# Cell 6: Missing values handling and winsorization (tailored to your dataset)
import numpy as np
from sklearn.preprocessing import MinMaxScaler
import joblib

# Identify identifier columns
id_cols = [c for c in ['planet_name', 'pl_name', 'name', 'rowid'] if c in df.columns]
print("Identifier columns:", id_cols)

# Build numeric column list: prefer canonical expected, else use detected numeric_like
expected_numeric_all = [
    'radius', 'mass', 'density', 'equilibrium_temperature',
    'orbital_period', 'semi_major_axis', 'distance',
    'host_star_temperature', 'host_star_luminosity', 'metallicity'
]
numeric_cols_for_drop = [c for c in expected_numeric_all if c in df.columns]
if not numeric_cols_for_drop:
    # fallback to previously detected numeric_like list
    numeric_cols_for_drop = fallback_numeric.copy()
print("Numeric columns to use:", numeric_cols_for_drop)

# Drop rows that are all-NaN across these numeric columns
if numeric_cols_for_drop:
    before = len(df)
    df = df.dropna(how='all', subset=numeric_cols_for_drop)
    after = len(df)
    print(f"Dropped {before-after} rows with all-NaN in numeric columns")
else:
    print("No numeric columns available to drop by subset")

# Winsorize numeric columns at 1st and 99th percentiles
def winsorize_series(s, lower_q=0.01, upper_q=0.99):
    if s.dropna().empty:
        return s
    lower = s.quantile(lower_q)
    upper = s.quantile(upper_q)
    return s.clip(lower, upper)

for col in numeric_cols_for_drop:
    df[col] = winsorize_series(df[col])

# Impute remaining numeric NaNs with median
for col in numeric_cols_for_drop:
    median_val = df[col].median()
    df[col] = df[col].fillna(median_val)

# Ensure host_star_type exists and fill
if 'host_star_type' in df.columns:
    df['host_star_type'] = df['host_star_type'].astype(str).replace({'nan':'UNKNOWN','None':'UNKNOWN'}).fillna('UNKNOWN')
else:
    print("host_star_type not present; continuing without it")
```

```python
# Cell 7 (updated): feature engineering - HSI, SCI, combined score (safe continuation)
from sklearn.preprocessing import MinMaxScaler
import joblib

# Confirm identifier and numeric columns
id_cols = [c for c in ['planet_name','rowid','pl_name','name'] if c in df.columns]
print("Identifier columns:", id_cols)

# Use canonical numeric columns present (you reported these exist)
canonical_numeric = ['radius','mass','density','equilibrium_temperature',
                     'orbital_period','semi_major_axis','distance',
                     'host_star_temperature','host_star_luminosity','metallicity']
numeric_available = [c for c in canonical_numeric if c in df.columns]
print("Canonical numeric available:", numeric_available)

# HSI selection: prefer the canonical order but only include available, exclude identifiers
hsi_priority = ['radius','equilibrium_temperature','mass','semi_major_axis','orbital_period','density']
hsi_cols = [c for c in hsi_priority if c in numeric_available]
print("HSI columns used:", hsi_cols)

# Scale HSI inputs (safe coercion and median fill)
if hsi_cols:
    hsi_df = df[hsi_cols].copy()
    for c in hsi_cols:
        hsi_df[c] = pd.to_numeric(hsi_df[c], errors='coerce').fillna(hsi_df[c].median())
    scaler_hsi = MinMaxScaler()
    hsi_scaled = scaler_hsi.fit_transform(hsi_df)
    joblib.dump(scaler_hsi, os.path.join(ARTIFACTS_DIR, "scaler_hsi.pkl"))
else:
    hsi_scaled = np.zeros((len(df),1))

# HSI: mid-range closeness score
def compute_hsi_row(scaled_row, cols):
    if len(cols) == 0:
        return 0.0
    score = 0.0
    w_total = 0.0
    # equal weights by default; tune later if needed
    for i, col in enumerate(cols):
        val = scaled_row[i]
        w = 1.0
        score += (1 - abs(val - 0.5)) * w
        w_total += w
    return score / (w_total if w_total else 1.0)

if hsi_cols:
    hsi_scores = np.apply_along_axis(lambda r: compute_hsi_row(r, hsi_cols), 1, hsi_scaled)
    df['habitability_index'] = MinMaxScaler().fit_transform(hsi_scores.reshape(-1,1)).ravel()
else:
    df['habitability_index'] = 0.0

# Stellar Compatibility Index (SCI)
sci_cols = [c for c in ['host_star_temperature','host_star_luminosity','metallicity','host_star_type'] if c in df.columns]
print("SCI columns used:", sci_cols)

def compute_sci(row):
    score = 0.0
    w = 0.0
    if 'host_star_temperature' in sci_cols:
        t = row['host_star_temperature']
        if pd.notna(t):
            score += max(0, 1 - abs(t - 5200)/3000)
            w += 1.0
    if 'host_star_luminosity' in sci_cols:
        L = row['host_star_luminosity']
        if pd.notna(L):
            score += max(0, 1 - abs(L - 1)/3)
            w += 1.0
    if 'metallicity' in sci_cols:
        m = row['metallicity']
        if pd.notna(m):
            score += max(0, 1 - abs(m - 0.0)/1.5)
            w += 0.6
    if 'host_star_type' in sci_cols:
        st = str(row.get('host_star_type','')).upper()
        if any(x in st for x in ['G','K']):
            score += 0.8
            w += 0.8
    return (score / w) if w else 0.0

if sci_cols:
    df['stellar_compatibility'] = df.apply(compute_sci, axis=1)
    df['stellar_compatibility'] = MinMaxScaler().fit_transform(df['stellar_compatibility'].values.reshape(-1,1)).ravel()
else:
    df['stellar_compatibility'] = 0.0

# Combined habitability score
w_hsi = 0.6
w_sci = 0.4
df['combined_habitability_score'] = (w_hsi * df['habitability_index'] + w_sci * df['stellar_compatibility'])

# Preview
display(df[[c for c in ['planet_name'] + hsi_cols + ['habitability_index','stellar_compatibility','combined_habitability_score'] if c in df.columns]].head(8))
```

```python
# Validation checks (run before Cell 8)
# 1. Check HSI/SCI ranges and NaNs
print("HSI min/max/na:", df['habitability_index'].min(), df['habitability_index'].max(), df['habitability_index'].isna().sum())
print("SCI min/max/na:", df['stellar_compatibility'].min(), df['stellar_compatibility'].max(), df['stellar_compatibility'].isna().sum())
print("Combined score min/max/na:", df['combined_habitability_score'].min(), df['combined_habitability_score'].max(), df['combined_habitability_score'].isna().sum())

# 2. Check for constant columns (zero variance) among numeric candidates
numeric_candidates = [c for c in ['radius','mass','density','equilibrium_temperature','orbital_period','semi_major_axis','distance','host_star_temperature','host_star_luminosity','metallicity'] if c in df.columns]
const_cols = [c for c in numeric_candidates if df[c].nunique(dropna=True) <= 1]
print("Constant numeric columns (consider dropping):", const_cols)

# 3. Confirm no identifier columns are in numeric_features later
print("Identifiers present:", [c for c in ['planet_name','rowid','pl_name'] if c in df.columns])

# 4. Quick target sanity (if you already created synthetic target earlier)
if 'target_habitable' in df.columns:
    print("Existing target distribution:\n", df['target_habitable'].value_counts())
```

```python
# Cell 8: create target variable (robust)
if 'habitability' in df.columns:
    df['target_habitable'] = df['habitability'].apply(lambda x: 1 if str(x).strip().lower() in ['habitable','yes','1','true'] else 0)
else:
    # Use combined score top percentile as synthetic label; adjust percentile if needed
    if 'combined_habitability_score' in df.columns and df['combined_habitability_score'].notna().any():
        percentile = 0.95  # change to 0.90 or 0.99 as you prefer
        threshold = df['combined_habitability_score'].quantile(percentile)
        df['target_habitable'] = (df['combined_habitability_score'] >= threshold).astype(int)
        print(f"Synthetic target created using top {int(percentile*100)}% threshold = {threshold:.4f}")
    else:
        df['target_habitable'] = 0
        print("Warning: No combined score available; target set to 0 for all rows")

# Show distribution
print("Target distribution:")
print(df['target_habitable'].value_counts())
```

```python
# Cell 9 (replace previous Cell 9): final feature selection and version-safe preprocessing
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Exclude identifiers and known low-value prefixes
exclude_exact = {'planet_name','pl_name','rowid','name'}
exclude_prefixes = ('gaia','tic','hip','hd','default_flag','flag','id')

# Canonical numeric list (only keep those present and non-constant)
canonical_numeric = ['radius','mass','density','equilibrium_temperature',
                     'orbital_period','semi_major_axis','distance',
                     'host_star_temperature','host_star_luminosity','metallicity']
numeric_features = [c for c in canonical_numeric if c in df.columns and df[c].nunique(dropna=True) > 1]

# Allow a small set of categorical features if present
categorical_allowed = [c for c in ['hostname','discoverymethod','disc_facility','disc_telescope','disc_instrument','soltype'] if c in df.columns]
categorical_features = categorical_allowed

# Build final features list and ensure no identifiers or ID-like prefixes are included
features = []
for c in numeric_features + categorical_features:
    low = c.lower()
    if c in exclude_exact:
        continue
    if any(low.startswith(p) for p in exclude_prefixes):
        continue
    features.append(c)

print("Numeric features kept:", numeric_features)
print("Categorical features kept:", categorical_features)
print("Final features list (count={}):".format(len(features)), features)

# Preprocessing pipelines (version-safe OneHotEncoder)
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Build OneHotEncoder kwargs in a version-agnostic way
ohe_kwargs = {}
try:
    # test new API
    OneHotEncoder(sparse_output=False)
    ohe_kwargs['sparse_output'] = False
except TypeError:
    ohe_kwargs['sparse'] = False

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='UNKNOWN')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', **ohe_kwargs))
])

preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
], remainder='drop')

print("Preprocessor created successfully.")
```

```python
# Features-definition cell: finalize features list, remove identifiers/IDs/flags, save features.txt
import os
ARTIFACTS_DIR = r"D:\Exo_Habit-AI\artifacts"
os.makedirs(ARTIFACTS_DIR, exist_ok=True)

# Rebuild numeric_features and categorical_features if not present
if 'numeric_features' not in globals() or 'categorical_features' not in globals():
    canonical_numeric = ['radius','mass','density','equilibrium_temperature',
                         'orbital_period','semi_major_axis','distance',
                         'host_star_temperature','host_star_luminosity','metallicity']
    numeric_features = [c for c in canonical_numeric if c in df.columns and df[c].nunique(dropna=True) > 1]
    categorical_allowed = [c for c in ['hostname','discoverymethod','disc_facility','disc_telescope','disc_instrument','soltype'] if c in df.columns]
    categorical_features = categorical_allowed

# Exclude identifiers and ID-like prefixes
exclude_exact = {'planet_name','pl_name','name','rowid'}
exclude_prefixes = ('gaia','tic','hip','hd','default_flag','flag','id')

# Build final features list
features = []
for c in numeric_features + categorical_features:
    low = c.lower()
    if c in exclude_exact:
        continue
    if any(low.startswith(p) for p in exclude_prefixes):
        continue
    features.append(c)

# Final guard: remove any accidental identifiers
for bad in ['planet_name','rowid','pl_name','name']:
    if bad in features:
        features.remove(bad)

# Persist features list for backend
with open(os.path.join(ARTIFACTS_DIR, "features.txt"), "w", encoding="utf-8") as f:
    for feat in features:
        f.write(feat + "\n")

print("Final features (count={}):".format(len(features)))
print(features)
print("Saved features.txt to", os.path.join(ARTIFACTS_DIR, "features.txt"))
```

```python
# Cell 10: train/test split and class weight calculation
from sklearn.model_selection import train_test_split
from sklearn.utils import class_weight

# Ensure features variable exists
assert 'features' in globals(), "Run the features-definition cell first."

X = df[features].copy()
y = df['target_habitable'].copy()

# Sanity check: target distribution
print("Target distribution before split:")
print(y.value_counts(dropna=False))

# Warn if single-class target
if y.nunique() < 2:
    print("Warning: target has less than 2 classes. Supervised training will not be meaningful until labels are provided.")
    
# Stratify only if there are at least 2 classes
stratify_arg = y if y.nunique() > 1 else None
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=stratify_arg, random_state=RANDOM_STATE)

# Compute class weights if applicable
class_weights = None
if y_train.nunique() > 1:
    cw = class_weight.compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
    class_weights = {cls: w for cls, w in zip(np.unique(y_train), cw)}
print("Class weights:", class_weights)
print("Train shape:", X_train.shape, "Test shape:", X_test.shape)
```

```python
# Cell 11: model pipelines and grid search training
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV, StratifiedKFold
import joblib

# Define models with safe defaults
models = {
    'random_forest': RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE, n_jobs=-1,
                                           class_weight='balanced' if class_weights else None),
    'xgboost': XGBClassifier(n_estimators=200, use_label_encoder=False, eval_metric='logloss', random_state=RANDOM_STATE),
    'logistic': LogisticRegression(max_iter=1000, random_state=RANDOM_STATE,
                                   class_weight='balanced' if class_weights else None)
}

# Build pipelines (preprocessor must be defined from Cell 9)
pipelines = {name: Pipeline(steps=[('preprocessor', preprocessor), ('clf', clf)]) for name, clf in models.items()}

# Cross-validation setup
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE) if y_train.nunique()>1 else None
best_models = {}

for name, pipe in pipelines.items():
    print("\nTraining:", name)
    if y_train.nunique() < 2:
        print("Skipping training because target has single class.")
        continue
    # Small grid for speed; expand later if needed
    if name == 'random_forest':
        param_grid = {'clf__n_estimators': [100, 200], 'clf__max_depth': [None, 10]}
    elif name == 'xgboost':
        param_grid = {'clf__n_estimators': [100, 200], 'clf__max_depth': [3, 6]}
    else:
        param_grid = {'clf__C': [0.1, 1.0, 10.0]}
    gs = GridSearchCV(pipe, param_grid, cv=cv, scoring='roc_auc', n_jobs=-1, verbose=1)
    gs.fit(X_train, y_train)
    print("Best params for", name, ":", gs.best_params_)
    best_models[name] = gs.best_estimator_
    # Save each best pipeline
    joblib.dump(gs.best_estimator_, os.path.join(ARTIFACTS_DIR, f"{name}_best.pkl"))
    print("Saved", name, "pipeline to artifacts.")
```

```python
# Cell 12: evaluate models and select best by AUC
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix

eval_results = {}
if not best_models:
    raise RuntimeError("No trained models found. Ensure training (Cell 11) completed successfully.")

for name, model in best_models.items():
    print("\nEvaluating model:", name)
    y_pred = model.predict(X_test)
    if hasattr(model, "predict_proba"):
        y_proba = model.predict_proba(X_test)[:,1]
    else:
        try:
            y_proba = model.decision_function(X_test)
        except:
            y_proba = np.zeros(len(y_test))
    auc = roc_auc_score(y_test, y_proba) if len(np.unique(y_test))>1 else None
    print("AUC:", auc)
    print("Classification report:")
    print(classification_report(y_test, y_pred))
    print("Confusion matrix:")
    print(confusion_matrix(y_test, y_pred))
    eval_results[name] = {'auc': auc}

# Select best model by AUC (fallback to first model if AUCs are None)
best_name = max(eval_results.items(), key=lambda x: (x[1]['auc'] if x[1]['auc'] is not None else -1))[0]
best_model = best_models[best_name]
print("\nSelected best model:", best_name, "with AUC:", eval_results[best_name]['auc'])
# Save selected pipeline
joblib.dump(best_model, os.path.join(ARTIFACTS_DIR, "best_pipeline.pkl"))
print("Saved best_pipeline.pkl to", ARTIFACTS_DIR)
```

```python
# SHAP and feature importances (run only if best_model exists)
import shap
if 'best_model' in globals():
    preprocessor.fit(X_train)
    num_cols = numeric_features
    cat_cols = []
    if categorical_features:
        ohe = preprocessor.named_transformers_['cat'].named_steps['onehot']
        cat_cols = list(ohe.get_feature_names_out(categorical_features))
    feature_names = num_cols + cat_cols

    clf = best_model.named_steps['clf']
    if hasattr(clf, 'feature_importances_'):
        fi = pd.Series(clf.feature_importances_, index=feature_names).sort_values(ascending=False)
        plt.figure(figsize=(8,6))
        fi.head(20).plot(kind='barh')
        plt.gca().invert_yaxis()
        plt.title("Top feature importances")
        plt.tight_layout()
        plt.savefig(os.path.join(ARTIFACTS_DIR, "feature_importances.png"), dpi=150)
        plt.show()

    try:
        X_sample = X_train.sample(min(200, len(X_train)), random_state=RANDOM_STATE)
        X_sample_proc = preprocessor.transform(X_sample)
        explainer = shap.Explainer(clf, X_sample_proc)
        X_test_proc = preprocessor.transform(X_test.sample(min(200, len(X_test)), random_state=RANDOM_STATE))
        shap_values = explainer(X_test_proc)
        shap.summary_plot(shap_values, features=X_test_proc, feature_names=feature_names, show=True)
    except Exception as e:
        print("SHAP explanation skipped or failed:", e)
else:
    print("No best_model available for explainability.")
```

```python
# Diagnostics utilities and sanitization
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.io as pio

ARTIFACTS_DIR = r"D:\Exo_Habit-AI\artifacts"
os.makedirs(ARTIFACTS_DIR, exist_ok=True)

# Replace infinite values with NaN and record counts
def sanitize_numeric(df):
    inf_mask = np.isinf(df.select_dtypes(include=[np.number])).any(axis=1)
    if inf_mask.any():
        print(f"Found {inf_mask.sum()} rows with infinite numeric values; replacing with NaN.")
    df.replace([np.inf, -np.inf], np.nan, inplace=True)
    return df

df = sanitize_numeric(df)

# Helper: ensure feature lists exist
if 'features' not in globals():
    # fallback: use numeric canonical features
    features = [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c]) and c not in ['rowid']]
numeric_features = [c for c in features if pd.api.types.is_numeric_dtype(df[c])]
categorical_features = [c for c in features if not pd.api.types.is_numeric_dtype(df[c])]
print("Numeric features for diagnostics:", numeric_features)
print("Categorical features for diagnostics:", categorical_features)
```

```python
# Missing values summary and bar chart
missing = df[features].isna().sum().sort_values(ascending=False)
missing_pct = (missing / len(df)).sort_values(ascending=False)

missing_df = pd.DataFrame({'n_missing': missing, 'pct_missing': missing_pct})
display(missing_df[missing_df['n_missing']>0].head(50))

# Bar chart for top missing features
top_missing = missing_df[missing_df['n_missing']>0].head(30)
plt.figure(figsize=(10,6))
sns.barplot(x=top_missing['pct_missing'].values, y=top_missing.index, palette='magma')
plt.xlabel('Fraction missing')
plt.title('Top features by fraction missing')
plt.tight_layout()
plt.savefig(os.path.join(ARTIFACTS_DIR, "missing_values_bar.png"), dpi=150)
plt.show()
```

```python
# Correlation heatmap for numeric features (use pairwise complete observations)
num_cols = [c for c in numeric_features if df[c].nunique(dropna=True) > 1]
if num_cols:
    corr = df[num_cols].corr(method='pearson')
    plt.figure(figsize=(12,10))
    sns.heatmap(corr, annot=False, cmap='coolwarm', center=0, linewidths=0.5)
    plt.title('Pearson Correlation Matrix (numeric features)')
    plt.tight_layout()
    plt.savefig(os.path.join(ARTIFACTS_DIR, "correlation_heatmap.png"), dpi=150)
    plt.show()

    # Save correlation table CSV
    corr.to_csv(os.path.join(ARTIFACTS_DIR, "feature_correlations.csv"))
    display(corr.round(3).iloc[:20,:20])
else:
    print("No numeric columns available for correlation heatmap.")
```

```python
# Pairplot (sampled to avoid heavy plotting)
sample_size = min(500, len(df))
sample_df = df[num_cols].sample(sample_size, random_state=RANDOM_STATE) if num_cols else pd.DataFrame()
if not sample_df.empty:
    sns.pairplot(sample_df, diag_kind='kde', plot_kws={'s':10, 'alpha':0.6})
    plt.suptitle('Pairwise relationships (sample)', y=1.02)
    plt.savefig(os.path.join(ARTIFACTS_DIR, "pairplot_sample.png"), dpi=150)
    plt.show()

    # Interactive Plotly scatter matrix (limited to top 8 numeric features by variance)
    var_sorted = df[num_cols].var().sort_values(ascending=False)
    top_vars = var_sorted.index.tolist()[:8]
    if top_vars:
        fig = px.scatter_matrix(df[top_vars].sample(min(1000, len(df))), dimensions=top_vars, title="Interactive scatter matrix (top variance features)")
        pio.write_html(fig, file=os.path.join(ARTIFACTS_DIR, "scatter_matrix_top_vars.html"), auto_open=False)
        display(fig)
else:
    print("Pairplot skipped: not enough numeric data.")
```

```python
# Distribution plots for numeric features: histogram + boxplot per feature (grid)
import math
plot_cols = num_cols
n = len(plot_cols)
if n:
    cols = 3
    rows = math.ceil(n/cols)
    plt.figure(figsize=(cols*5, rows*3.5))
    for i, col in enumerate(plot_cols):
        plt.subplot(rows, cols, i+1)
        sns.histplot(df[col].dropna(), bins=40, kde=True, color='steelblue')
        plt.title(col)
    plt.tight_layout()
    plt.savefig(os.path.join(ARTIFACTS_DIR, "numeric_histograms.png"), dpi=150)
    plt.show()

    # Boxplots to highlight outliers
    plt.figure(figsize=(cols*5, rows*1.5))
    for i, col in enumerate(plot_cols):
        plt.subplot(1, cols, (i % cols)+1)
        sns.boxplot(x=df[col].dropna(), color='lightcoral')
        plt.title(col)
        if (i+1) % cols == 0 and i+1 < n:
            plt.show()
    plt.tight_layout()
    plt.savefig(os.path.join(ARTIFACTS_DIR, "numeric_boxplots.png"), dpi=150)
    plt.show()
else:
    print("No numeric features to plot distributions.")
```

```python
# Outlier detection using IQR and Z-score
from scipy import stats

outlier_summary = []
for col in num_cols:
    series = df[col].dropna()
    if series.empty:
        continue
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    iqr_outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    # z-score outliers (abs z > 3)
    z_scores = np.abs(stats.zscore(series))
    z_outliers = (z_scores > 3).sum()
    outlier_summary.append({'feature': col, 'iqr_outliers': int(iqr_outliers), 'z_outliers': int(z_outliers), 'n_nonnull': int(series.count())})

outlier_df = pd.DataFrame(outlier_summary).sort_values(by='iqr_outliers', ascending=False)
display(outlier_df.head(30))
outlier_df.to_csv(os.path.join(ARTIFACTS_DIR, "outlier_summary.csv"), index=False)

# Show top 5 rows flagged as outliers for the top 3 features by IQR outliers
top_features = outlier_df['feature'].head(3).tolist()
for f in top_features:
    q1 = df[f].quantile(0.25)
    q3 = df[f].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    out_rows = df[(df[f] < lower) | (df[f] > upper)][[ 'planet_name', f ] + ([ 'combined_habitability_score'] if 'combined_habitability_score' in df.columns else [])].head(10)
    print(f"\nTop outlier rows for {f}:")
    display(out_rows)
```

```python
# Robust categorical summaries: handle empty categorical list safely
cat_cols = [c for c in df.columns if df[c].dtype == 'object' or pd.api.types.is_categorical_dtype(df[c])]
# Limit to the categorical features we care about (if any)
cat_cols = [c for c in cat_cols if c in (categorical_features if 'categorical_features' in globals() else []) or c in ['planet_name','hostname','discoverymethod','host_star_type']]

print("Categorical columns considered:", cat_cols)

cat_summary = {}
for c in cat_cols:
    series = df[c].fillna('<<NA>>').astype(str)
    vc = series.value_counts(dropna=False).head(10)
    unknown_like = series.str.upper().isin(['UNKNOWN','NAN','NONE','']).sum()
    cat_summary[c] = {
        'n_unique': int(df[c].nunique(dropna=True)),
        'n_missing': int(df[c].isna().sum()),
        'n_unknown_like': int(unknown_like),
        'top_values': vc.to_dict()
    }

if cat_summary:
    cat_summary_df = pd.DataFrame.from_dict(cat_summary, orient='index')
    # Only select columns that exist
    cols_to_show = [c for c in ['n_unique','n_missing','n_unknown_like'] if c in cat_summary_df.columns]
    display(cat_summary_df[cols_to_show].sort_values(by='n_missing', ascending=False).head(50))
    # Save top categories to file
    with open(os.path.join(ARTIFACTS_DIR, "categorical_summary.txt"), "w", encoding="utf-8") as f:
        for c, info in cat_summary.items():
            f.write(f"Column: {c}\n")
            f.write(f"  n_unique: {info['n_unique']}\n")
            f.write(f"  n_missing: {info['n_missing']}\n")
            f.write(f"  n_unknown_like: {info['n_unknown_like']}\n")
            f.write(f"  top_values: {info['top_values']}\n\n")
    print("Categorical summary saved to artifacts.")
else:
    print("No categorical columns found to summarize.")
```

```python
# Finalize features and save features.txt
ARTIFACTS_DIR = r"D:\Exo_Habit-AI\artifacts"
os.makedirs(ARTIFACTS_DIR, exist_ok=True)

if 'numeric_features' not in globals() or 'categorical_features' not in globals():
    canonical_numeric = ['radius','mass','density','equilibrium_temperature','orbital_period','semi_major_axis','distance','host_star_temperature','host_star_luminosity','metallicity']
    numeric_features = [c for c in canonical_numeric if c in df.columns and df[c].nunique(dropna=True) > 1]
    categorical_allowed = [c for c in ['hostname','discoverymethod','disc_facility','disc_telescope','disc_instrument','soltype'] if c in df.columns]
    categorical_features = categorical_allowed

exclude_exact = {'planet_name','pl_name','name','rowid'}
exclude_prefixes = ('gaia','tic','hip','hd','default_flag','flag','id')

features = []
for c in numeric_features + categorical_features:
    low = c.lower()
    if c in exclude_exact: continue
    if any(low.startswith(p) for p in exclude_prefixes): continue
    features.append(c)

for bad in ['planet_name','rowid','pl_name','name']:
    if bad in features:
        features.remove(bad)

with open(os.path.join(ARTIFACTS_DIR, "features.txt"), "w", encoding="utf-8") as f:
    for feat in features:
        f.write(feat + "\n")

print("Final features (count={}):".format(len(features)))
print(features)
print("Saved features.txt to", os.path.join(ARTIFACTS_DIR, "features.txt"))
```
