## 🪐 ExoHabitAI – Module 1
### Data Collection & Management

#### 📌 Cell 1: Imports & Path Configuration

```python
import pandas as pd
import numpy as np
import os
import json
from datetime import datetime
```

#### 📌 Cell 2: Define Project Paths

```python
BASE_PATH = r"D:\ExoHabit-AI"

RAW_DATA_PATH = os.path.join(BASE_PATH, "data", "raw")
INTERIM_DATA_PATH = os.path.join(BASE_PATH, "data", "interim")
PROFILE_PATH = os.path.join(BASE_PATH, "data", "data_profile")
LOG_PATH = os.path.join(BASE_PATH, "logs")

RAW_FILE = r"D:\ExoHabit-AI\ExoPlanet.csv"
```

#### 📌 Cell 3: Create Required Folder Structure

```python
for path in [RAW_DATA_PATH, INTERIM_DATA_PATH, PROFILE_PATH, LOG_PATH]:
    os.makedirs(path, exist_ok=True)

print("✅ Project directory structure verified")
```

#### 📌 Cell 4: Load Dataset

```python
df_raw = pd.read_csv(RAW_FILE, dtype={"hd_name": str, "hip_name": str})

print("Dataset Loaded Successfully")
print("Shape:", df_raw.shape)
df_raw.head()
```

#### 📌 Cell 5: Define Required Schema

```python
EXPECTED_COLUMNS = [
    "pl_name",        # Planet Name
    "pl_rade",        # Planet Radius (Earth radii)
    "pl_masse",       # Planet Mass (Earth mass)
    "pl_orbper",      # Orbital Period (days)
    "pl_eqt",         # Equilibrium Temperature (K)
    "st_teff",        # Stellar Temperature (K)
    "st_rad",         # Stellar Radius (Solar radii)
    "st_mass",        # Stellar Mass (Solar mass)
    "st_lum"          # Stellar Luminosity
]
```

#### 📌 Cell 6: Schema Validation

```python
missing_cols = set(EXPECTED_COLUMNS) - set(df_raw.columns)

if missing_cols:
    raise ValueError(f"❌ Missing required columns: {missing_cols}")

print("✅ Schema validation passed")
```

#### 📌 Cell 7: Type Conversion & Standardization

```python
df_validated = df_raw[EXPECTED_COLUMNS].copy()

for col in EXPECTED_COLUMNS:
    if col != "pl_name":
        df_validated[col] = pd.to_numeric(df_validated[col], errors="coerce")

print("✅ Data types standardized")
df_validated.dtypes
```

#### 📌 Cell 8: Basic Data Quality Checks

```python
quality_report = {
    "total_rows": len(df_validated),
    "missing_values": df_validated.isnull().sum().to_dict(),
    "duplicate_rows": int(df_validated.duplicated().sum())
}

quality_report
```

#### 📌 Cell 9: Save Validated Dataset

```python
validated_path = os.path.join(
    INTERIM_DATA_PATH,
    "exoplanets_validated.csv"
)

df_validated.to_csv(validated_path, index=False)

print(f"✅ Validated dataset saved to:\n{validated_path}")
```

#### 📌 Cell 10: Generate Data Profiling Report (JSON)

```python
profile = {
    "generated_on": datetime.now().isoformat(),
    "rows": len(df_validated),
    "columns": list(df_validated.columns),
    "missing_values": df_validated.isnull().sum().to_dict(),
    "numeric_summary": df_validated.describe().to_dict()
}

profile_file = os.path.join(PROFILE_PATH, "profile_report.json")

with open(profile_file, "w") as f:
    json.dump(profile, f, indent=4)

print(f"✅ Data profiling report saved to:\n{profile_file}")
```

#### 📌 Cell 11: Final Verification

```python
print("Module 1 Completed Successfully 🚀")
print("Next Output Files:")
print("• exoplanets_validated.csv")
print("• profile_report.json")
```

## 🪐 ExoHabitAI – Module 2
### Data Cleaning & Feature Engineering
#### 📌 Cell 1: Imports & Paths

```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

#### 📌 Cell 2: Define Paths

```python
BASE_PATH = r"D:\ExoHabit-AI"

INTERIM_PATH = os.path.join(BASE_PATH, "data", "interim")
PROCESSED_PATH = os.path.join(BASE_PATH, "data", "processed")

INPUT_FILE = os.path.join(INTERIM_PATH, "exoplanets_validated.csv")
OUTPUT_FILE = os.path.join(PROCESSED_PATH, "exoplanets_features_v1.csv")

os.makedirs(PROCESSED_PATH, exist_ok=True)
```

#### 📌 Cell 3: Load Validated Dataset

```python
df = pd.read_csv(INPUT_FILE)

print("Dataset Loaded")
print("Shape:", df.shape)
df.head()
```

#### 📌 Cell 4: Missing Value Analysis

```python
missing_percent = (df.isnull().sum() / len(df)) * 100
missing_percent.sort_values(ascending=False)
```

#### 📌 Cell 5: Astro-Aware Missing Value Imputation

##### Why this matters

###### Median is safer for skewed astrophysical data

###### Stellar values influence planetary estimates

```python
NUM_COLS = df.columns.drop("pl_name")

medians = df[NUM_COLS].median()
df[NUM_COLS] = df[NUM_COLS].fillna(medians)

print("✅ Missing values imputed using vectorized median strategy")
```

#### 📌 Cell 6: Outlier Handling (IQR – conservative)

```python
def cap_outliers_iqr(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return np.clip(series, lower, upper)

for col in NUM_COLS:
    df[col] = cap_outliers_iqr(df[col])

print("✅ Outliers capped using IQR method")
```

#### 📌 Cell 7: Core Astro Feature Engineering
##### 1️⃣ Planet Density (relative proxy)

```python
df["planet_density_index"] = df["pl_masse"] / (df["pl_rade"] ** 3)
```

##### 2️⃣ Stellar Flux (relative to Earth)

```python
df["stellar_flux"] = df["st_lum"] / (df["st_rad"] ** 2)
```

##### 3️⃣ Habitable Zone Distance (simplified)

```python
# Enforce physical constraint: luminosity must be positive
df["st_lum"] = df["st_lum"].clip(lower=1e-6)

# Habitable zone distance (safe)
df["habitable_zone_distance"] = np.sqrt(df["st_lum"])

# Defensive fill (no inplace)
hz_median = df["habitable_zone_distance"].median()
df["habitable_zone_distance"] = df["habitable_zone_distance"].fillna(hz_median)
```

#### 📌 Cell 8: Habitability Score Index (HSI)

##### Astrophysics-inspired heuristic (0–1 scale)

```python
def habitability_score(row):
    score = 0.0

    # Radius score (Earth ~1 R⊕)
    if pd.notna(row["pl_rade"]):
        score += max(
            0,
            1 - abs(row["pl_rade"] - 1.0) / 1.5
        ) * 0.25

    # Mass score (Earth ~1 M⊕)
    if pd.notna(row["pl_masse"]):
        score += max(
            0,
            1 - abs(row["pl_masse"] - 1.0) / 5.0
        ) * 0.25

    # Temperature score (Earth ~288 K)
    if pd.notna(row["pl_eqt"]):
        score += max(
            0,
            1 - abs(row["pl_eqt"] - 288) / 150
        ) * 0.25

    # Stellar flux score (Earth ~1)
    if pd.notna(row["stellar_flux"]):
        score += max(
            0,
            1 - abs(row["stellar_flux"] - 1.0) / 2.0
        ) * 0.25

    return round(score, 3)


df["habitability_score_index"] = df.apply(habitability_score, axis=1)

print("✅ Habitability Score Index computed (continuous, data-driven)")
```

#### 📌 Cell 9: Stellar Compatibility Index (SCI)

##### Measures host star suitability.

```python
def stellar_compatibility(row):
    score = 0.0

    # Star temperature: late K to G-type
    if pd.notna(row["st_teff"]) and 4800 <= row["st_teff"] <= 6000:
        score += 0.5

    # Stellar mass stability (Sun-like)
    if pd.notna(row["st_mass"]) and 0.8 <= row["st_mass"] <= 1.2:
        score += 0.5

    return score


df["stellar_compatibility_index"] = df.apply(stellar_compatibility, axis=1)

print("✅ Stellar Compatibility Index computed")
```

#### 📌 Cell 10: Label Generation (Target Variable)

##### Binary classification: Habitable vs Not

```python
HSI_PERCENTILE = 0.93

hsi_threshold = df["habitability_score_index"].quantile(HSI_PERCENTILE)

df["is_habitable"] = (
    (df["habitability_score_index"] >= hsi_threshold) &
    (df["stellar_compatibility_index"] >= 0.5)
).astype(int)

df["is_habitable"].value_counts()
```

```python
print(f"HSI threshold (@{int(HSI_PERCENTILE*100)}th percentile): {hsi_threshold:.3f}")
```

#### 📌 Cell 11: Feature Normalization

```python
# Explicitly define feature columns (no target leakage)
FEATURE_COLS = df.columns.drop([
    "pl_name",
    "is_habitable"
])

print(f"✅ Feature columns locked ({len(FEATURE_COLS)} features)")
```

#### 📌 Cell 12: Data Quality Validation

```python
print("🔍 Dataset shape:", df.shape)

print("\n🔍 Missing values check:")
print(df.isnull().sum().sort_values(ascending=False).head())

print("\n🔍 Infinite values check:")
print(np.isinf(df[FEATURE_COLS]).sum().sum())

print("\n🔍 Target distribution:")
print(df["is_habitable"].value_counts(normalize=True))

df[FEATURE_COLS].describe().T
```

#### 📌 Cell 12A: SCI vs Target Consistency Check

```python
# Mean SCI by habitability class (expected to differ by construction)
sci_by_class = df.groupby("is_habitable")["stellar_compatibility_index"].mean()

print("🔍 Mean Stellar Compatibility Index by class:")
print(sci_by_class)
```

#### 📌 Cell 12B: SCI × Habitability Crosstab

```python
sci_crosstab = pd.crosstab(
    df["stellar_compatibility_index"],
    df["is_habitable"],
    normalize="index"
)

print("🔍 SCI vs Habitability (row-normalized):")
print(sci_crosstab)
```

#### 📌 Cell 12C: Zero-Variance Feature Detection

```python
# Identify zero-variance (constant) features
zero_variance_features = [
    col for col in FEATURE_COLS
    if df[col].nunique() <= 1
]

print("⚠️ Zero-variance features detected:")
print(zero_variance_features)
```

#### 📌 Cell 12D: Correlation Snapshot

```python
corr_with_target = (
    df[FEATURE_COLS]
    .assign(is_habitable=df["is_habitable"])
    .corr()["is_habitable"]
    .sort_values(ascending=False)
)

print("🔍 Feature correlation with target (diagnostic only):")
print(corr_with_target.head(10))
```

#### 📌 Cell 13: Visualization

```python
plt.figure(figsize=(6,4))
sns.countplot(x="is_habitable", data=df)
plt.title("Habitability Class Distribution")
plt.ylabel("Planet Count")
plt.xlabel("Is Habitable")
plt.show()
```

### 🪐 ExoHabitAI — MODULE 3
#### Machine Learning Dataset Preparation & Train/Test Strategy
#### 📌 Cell 1: Imports

```python
import pandas as pd
import numpy as np
import os
import json

from sklearn.model_selection import train_test_split
from sklearn.feature_selection import SelectKBest, f_classif
```

#### 📌 Cell 2: Define Paths

```python
BASE_PATH = r"D:\ExoHabit-AI"

PROCESSED_PATH = os.path.join(BASE_PATH, "data", "processed")
ML_PATH = os.path.join(BASE_PATH, "data", "ml")

INPUT_FILE = os.path.join(PROCESSED_PATH, "exoplanets_features_v1.csv")

os.makedirs(ML_PATH, exist_ok=True)
```

#### 📌 Cell 3: Load Module 2 Dataset

```python
df = pd.read_csv(INPUT_FILE)

print("✅ Dataset loaded")
print("Shape:", df.shape)
df.head()
```

####  📌 Cell 4: Define Target & Remove Non-ML Columns
##### 🔒 🚨 ABSOLUTE RULE
###### Any feature used to construct the label must NOT be used for ML.

```python
TARGET = "is_habitable"

LEAKAGE_COLS = [
    "habitability_score_index",
    "stellar_compatibility_index"
]

DROP_COLS = ["pl_name"] + LEAKAGE_COLS

X = df.drop(columns=[TARGET] + DROP_COLS)
y = df[TARGET]

print("✅ Feature matrix prepared")
print("Dropped leakage columns:", LEAKAGE_COLS)
print("X shape:", X.shape)
print("Target distribution:")
print(y.value_counts(normalize=True))
```

#### 📌 Cell 5: Drop Zero-Variance Features
##### 🚨 These MUST be removed before feature selection

```python
zero_var_cols = X.columns[X.nunique() <= 1].tolist()

if zero_var_cols:
    print("⚠️ Dropping zero-variance features:", zero_var_cols)
    X = X.drop(columns=zero_var_cols)
else:
    print("✅ No zero-variance features detected")
```

#### 📌 Cell 6: Feature Selection

```python
K = min(15, X.shape[1])  # tighter cap = better generalization

selector = SelectKBest(score_func=f_classif, k=K)
X_selected = selector.fit_transform(X, y)

selected_features = X.columns[selector.get_support()].tolist()

print(f"✅ Selected {len(selected_features)} features:")
selected_features
```

#### 📌 Cell 7: Persist Feature List

```python
with open(os.path.join(ML_PATH, "selected_features.json"), "w") as f:
    json.dump(selected_features, f, indent=4)

print("✅ Feature list saved")
```

#### 📌 Cell 8: Stratified Train / Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_selected,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
```

#### 📌 Cell 9: Class Imbalance Weight

```python
pos_weight = (y_train == 0).sum() / (y_train == 1).sum()

print("⚖️ scale_pos_weight (for XGBoost):", round(pos_weight, 2))
```

#### 📌 Cell 10: Save ML-Ready Arrays

```python
np.save(os.path.join(ML_PATH, "X_train.npy"), X_train)
np.save(os.path.join(ML_PATH, "X_test.npy"), X_test)
np.save(os.path.join(ML_PATH, "y_train.npy"), y_train)
np.save(os.path.join(ML_PATH, "y_test.npy"), y_test)

print("✅ ML datasets saved")
```

### 🪐 ExoHabitAI — MODULE 4
#### AI Model for Habitability Prediction (Core Intelligence)
#### 📌 Cell 1: Imports 

```python
import numpy as np
import pandas as pd
import os
import json

from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    precision_recall_curve
)
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import StackingClassifier

import xgboost as xgb
import joblib
```

#### 📌 Cell 2: Paths & Load ML Artifacts

```python
BASE_PATH = r"D:\ExoHabit-AI"

ML_PATH = os.path.join(BASE_PATH, "data", "ml")
MODEL_PATH = os.path.join(BASE_PATH, "models")

os.makedirs(MODEL_PATH, exist_ok=True)

X_train = np.load(os.path.join(ML_PATH, "X_train.npy"))
X_test = np.load(os.path.join(ML_PATH, "X_test.npy"))
y_train = np.load(os.path.join(ML_PATH, "y_train.npy"))
y_test = np.load(os.path.join(ML_PATH, "y_test.npy"))

with open(os.path.join(ML_PATH, "selected_features.json")) as f:
    selected_features = json.load(f)

print("✅ ML datasets loaded")
print("Train shape:", X_train.shape)
print("Selected features:", selected_features)
```

#### 📌 Cell 3: Class Imbalance Handling

```python
pos_weight = (y_train == 0).sum() / (y_train == 1).sum()
print("✅ scale_pos_weight:", round(pos_weight, 2))
```

#### 📌 Cell 4: Train Random Forest (Stability Anchor)

```python
rf_model = RandomForestClassifier(
    n_estimators=400,
    max_depth=14,
    min_samples_split=5,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)
print("✅ Random Forest trained")
```

#### 📌 Cell 5: Train XGBoost (Recall-Focused Signal Learner)

```python
xgb_model = xgb.XGBClassifier(
    n_estimators=500,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.85,
    colsample_bytree=0.85,
    scale_pos_weight=pos_weight,
    objective="binary:logistic",
    eval_metric="auc",
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(X_train, y_train)
print("✅ XGBoost trained")
```

#### 📌 Cell 6: Build Stacking Ensemble (Leakage-Safe)
##### Architecture

###### Base: RF + XGB

###### Meta: Logistic Regression (probability calibration)

```python
estimators = [
    ("rf", rf_model),
    ("xgb", xgb_model)
]

stacking_model = StackingClassifier(
    estimators=estimators,
    final_estimator=LogisticRegression(
        class_weight="balanced",
        max_iter=1000
    ),
    cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
    n_jobs=-1,
    passthrough=False
)

stacking_model.fit(X_train, y_train)
print("✅ Stacking model trained")
```

#### 📌 Cell 7: Evaluation Utility

```python
def evaluate_model(model, X, y, threshold=0.5):
    probs = model.predict_proba(X)[:, 1]
    preds = (probs >= threshold).astype(int)

    return {
        "accuracy": accuracy_score(y, preds),
        "precision": precision_score(y, preds, zero_division=0),
        "recall": recall_score(y, preds),
        "f1": f1_score(y, preds),
        "roc_auc": roc_auc_score(y, probs)
    }
```

#### 📌 Cell 8: Baseline Evaluation

```python
baseline_metrics = evaluate_model(
    stacking_model,
    X_test,
    y_test,
    threshold=0.5
)

print("🔍 Baseline metrics:")
baseline_metrics
```

#### 📌 Cell 9: Recall-Optimized Threshold Tuning

```python
# Recall-Optimized Threshold Tuning
probs = stacking_model.predict_proba(X_test)[:, 1]
from sklearn.metrics import precision_recall_curve
precision, recall, thresholds = precision_recall_curve(y_test, probs)

recall_target = 0.80
best_threshold = None

for r, t in zip(recall, thresholds):
    if r >= recall_target:
        best_threshold = t
        break

print("Best threshold for recall >= 0.8:", best_threshold)
```

```python
with open(os.path.join(MODEL_PATH, "optimal_threshold.json"), "w") as f:
    json.dump({"threshold": float(best_threshold)}, f, indent=4)

print("✅ Optimal threshold saved")
```

```python
optimized_metrics = evaluate_model(
    stacking_model, X_test, y_test, threshold=best_threshold
)
optimized_metrics
```
